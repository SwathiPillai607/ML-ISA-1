```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
```

Data file

```python
DATA_PATH = "data.csv"  
TEST_SIZE = 0.2
RANDOM_STATE = 42
```

Column names

```python
columns = [
    "id", "diagnosis",
    "radius_mean", "texture_mean", "perimeter_mean", "area_mean",
    "smoothness_mean", "compactness_mean", "concavity_mean",
    "concave_points_mean", "symmetry_mean", "fractal_dimension_mean",

    "radius_se", "texture_se", "perimeter_se", "area_se",
    "smoothness_se", "compactness_se", "concavity_se",
    "concave_points_se", "symmetry_se", "fractal_dimension_se",

    "radius_worst", "texture_worst", "perimeter_worst", "area_worst",
    "smoothness_worst", "compactness_worst", "concavity_worst",
    "concave_points_worst", "symmetry_worst", "fractal_dimension_worst"
]
```

Load dataset

```python
df = pd.read_csv(DATA_PATH, header=None, names=columns)

df.head()
```

Preprocessing

```python
# Drop ID column
df.drop("id", axis=1, inplace=True)

# Encode target
df["diagnosis"] = df["diagnosis"].map({
    "M": 1,
    "B": 0
})

df.head()
```

Feature and target

```python
X = df.drop("diagnosis", axis=1)
y = df["diagnosis"]
```

Train test split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y
)
```

Feature scaling

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

Logistic regression

```python
log_model = LogisticRegression(max_iter=1000)
log_model.fit(X_train, y_train)

log_train_pred = log_model.predict(X_train)
log_test_pred = log_model.predict(X_test)
```

Logistic regression results

```python
print("===== LOGISTIC REGRESSION =====\n")

train_error = 1 - accuracy_score(y_train, log_train_pred)
test_error = 1 - accuracy_score(y_test, log_test_pred)

print("Training Error :", train_error)
print("Test Error     :", test_error)
print("Generalization Gap:", test_error - train_error)

print("\nAccuracy :", accuracy_score(y_test, log_test_pred))
print("Precision:", precision_score(y_test, log_test_pred))
print("Recall   :", recall_score(y_test, log_test_pred))
print("F1-score :", f1_score(y_test, log_test_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, log_test_pred))
```

Decision tree

```python
tree_model = DecisionTreeClassifier(random_state=42)
tree_model.fit(X_train, y_train)

tree_train_pred = tree_model.predict(X_train)
tree_test_pred = tree_model.predict(X_test)
```

Decision tree results

```python
print("===== DECISION TREE =====\n")

train_error = 1 - accuracy_score(y_train, tree_train_pred)
test_error = 1 - accuracy_score(y_test, tree_test_pred)

print("Training Error :", train_error)
print("Test Error     :", test_error)
print("Generalization Gap:", test_error - train_error)

print("\nAccuracy :", accuracy_score(y_test, tree_test_pred))
print("Precision:", precision_score(y_test, tree_test_pred))
print("Recall   :", recall_score(y_test, tree_test_pred))
print("F1-score :", f1_score(y_test, tree_test_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, tree_test_pred))
```

### Conclusion: 
From the experimental results, logistic regression achieves a training error of 1.31% and a test error of 3.50%, resulting in a small generalization gap of approximately 2.19%. Since both training and test errors are low and close to each other, the model demonstrates good generalization ability. This indicates that logistic regression does not suffer from overfitting or underfitting and is able to learn meaningful patterns from the dataset while maintaining stable performance on unseen data.

In contrast, the decision tree model shows zero training error but a significantly higher test error of about 7.01%, producing a large generalization gap of 7.01%. This behavior clearly indicates overfitting, as the model memorizes the training data instead of learning general patterns. Although the decision tree performs well on training samples, its high variance reduces its ability to generalize to new data. Therefore, based on generalization performance and stability, logistic regression is the more reliable model for deployment in this breast cancer classification task.
